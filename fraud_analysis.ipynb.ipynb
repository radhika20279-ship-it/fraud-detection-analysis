```python
#What % of transactions are fraud?

fraud_pct = df['is_fraud'].value_counts(normalize = True) * 100
print(fraud_pct)
```

## Key Insight 1
Only 0.58% of transactions are fraudulent (7,506 out of 1,296,675).
This reflects real-world fraud rates and highlights the need for 
pattern detection rather than volume-based analysis.

```python
#Which category has the most fraud?

df[df['is_fraud'] ==1]['category'].value_counts()
```

## Key Insight 2
Top 3 fraud categories are grocery_pos (1,743), shopping_net (1,713) 
and misc_net (915). Online transactions account for a significant 
portion of fraud cases.

```python
#Average transaction amount: fraud vs normal?
df.groupby(['category', 'is_fraud'])['amt'].agg(['count', 'mean', 'sum']).round(2)
```

## Key Insight 3
Fraud transactions have much higher average amounts than normal ones.
Travel fraud averages $1,051 vs $111 for normal transactions (9x higher).
Shopping_net fraud averages $999 vs $72 for normal (13x higher).
High transaction amounts in these categories are a strong fraud indicator.

```python
#Which hour sees the most fraud?
df[df['is_fraud'] == 1]['hour'].value_counts().sort_index()
```

## Key Insight 4
Fraud is heavily concentrated in late night hours.
- 10PM and 11PM alone account for 3,835 fraud cases
- Midnight to 3AM adds another 2,527 cases
- Together 10PM–3AM = over 51% of all fraud
- Daytime hours (4AM–11AM) show minimal fraud activity
This suggests fraudsters deliberately act during low-monitoring hours.

```python
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)

print("Libraries ready!")
```

```python
# Chart 1 - Fraud by Category
fraud_by_category = df[df['is_fraud'] == 1]['category'].value_counts()

plt.figure(figsize=(12, 5))
sns.barplot(x=fraud_by_category.index, y=fraud_by_category.values, 
            hue=fraud_by_category.index, palette='Blues_r', legend=False)

plt.title('Fraud Cases by Transaction Category', fontsize=16)
plt.xlabel('Category', fontsize=12)
plt.ylabel('Number of Fraud Cases', fontsize=12)
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
# Chart 2 - Fraud by Hour of Day
fraud_by_hour = df[df['is_fraud'] == 1]['hour'].value_counts().sort_index()

plt.figure(figsize=(12, 5))
sns.barplot(x=fraud_by_hour.index, y=fraud_by_hour.values, 
            hue=fraud_by_hour.index, palette='Blues_r', legend=False)

plt.title('Fraud Cases by Hour of Day', fontsize=16)
plt.xlabel('Hour (0 = Midnight, 23 = 11PM)', fontsize=12)
plt.ylabel('Number of Fraud Cases', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# Chart 3 - Transaction Amount Fraud vs Normal
plt.figure(figsize=(12, 5))
sns.boxplot(x='is_fraud', y='amt', data=df, 
            hue='is_fraud', palette='Blues', legend=False)

plt.title('Transaction Amount: Fraud vs Normal', fontsize=16)
plt.xlabel('Is Fraud (0 = Normal, 1 = Fraud)', fontsize=12)
plt.ylabel('Transaction Amount ($)', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# Chart 4 - Fraud by Gender
fraud_by_gender = df[df['is_fraud'] == 1]['gender'].value_counts()

plt.figure(figsize=(8, 5))
sns.barplot(x=fraud_by_gender.index, y=fraud_by_gender.values,
            hue=fraud_by_gender.index, palette='Blues_r', legend=False)

plt.title('Fraud Cases by Gender', fontsize=16)
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Number of Fraud Cases', fontsize=12)

plt.tight_layout()
plt.show()
```

## Key Insight 5
Fraud cases are almost equally split between male (3,700) 
and female (3,600) customers. Gender alone is not a strong 
predictor of fraud.
